Tensors are ways to store an input data on which maths is done 
- it also stores the weights and biases 

in phy - tensors r used to represent quantities (scalars - mangnitude) (vectors - mangnitude and direction)

scalars - 0 tensors  , vector - 1D , Matrix - 2D , 3D and so on - N Dimensioanl

Mathematical operations are performed on these operations
GPUs/TPUs make these computations faster

NNs starts with randomn numbers on these tensors (on weights and biases) and then adjust based on data through learning process

important parametrs for tensor
1. - dtype - data type of tensor  (READ MORE LATER ABT PRECISION AND BITS)*
   - deals with precision (ex: float 16 less precise than float32)
   - smaller dtype uses less memory
   - some gpus are optmized for float16

2. device - can be CPU(cpu)/GPU(cuda)

3. - requires_grad - tracking gradients 
   - pytorch record operations on the tensor for backpropagtion
   - must be True for nn weights
   - i/p tensors for inference usually have False
